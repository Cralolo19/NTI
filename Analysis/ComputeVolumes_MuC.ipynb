```python
import numpy as np
from tabulate import tabulate

def Vol(det_part, det_geometry):
    # Function to compute the volume of various detector parts
    if det_part == "Nozzles":
        r1, r2, r3, l = det_geometry
        Volume = np.pi*l*( r3**2 + r3*r2 + r2**2 - 3*r1**2 )/3 # cm^3
        Volume = Volume*2e-6 # m^3
        return Volume
    if det_part == "ECal" or det_part == "HCal" or det_part == "Solenoid" or det_part == "MuonDetector":
        r1, r2, r3, l = det_geometry
        Volume = np.pi*l*( 3*r3**2 - r2**2 - r2*r1 - r1**2 )/3
        Volume = Volume*2e-6 # m^3
        return Volume

####################################################################################################################
########################################## Introducing the detector geometry #######################################
####################################################################################################################
Nozzles1 = [2.2, 2.2, 31, 230.7 - 6.5] # [r1, r2, r3, l]
Nozzles2 = [2.2, 31, 78.2, 563.8 - 230.7] # [r1, r2, r3, l]
ECal_Barrel = [150, 150, 170.2, 221 - 0] # [r1, r2, r3, l]
ECal_Endcap = [31, 33.9, 170, 250.9 - 230.7 ] # [r1, r2, r3, l]
HCal_Barrel = [174, 174, 333, 221 - 0] # [r1, r2, r3, l]
HCal_Endcap1 = [170, 170, 324.6, 250.9 - 235.4] # [r1, r2, r3, l]
HCal_Endcap2 = [33.9, 56.8, 324.6, 412.9 - 250.9] # [r1, r2, r3, l]
Solenoid_Inner = [348.3, 348.3, 352.3, 412.9 - 0] # [r1, r2, r3, l]
Solenoid_Middle = [364.9, 364.9, 399.3, 412.9 - 0] # [r1, r2, r3, l]
Solenoid_Outer = [425, 425, 429, 412.9 - 0] # [r1, r2, r3, l]
MuonDetector_Barrel = [446.1, 446.1, 645, 417.9 - 0] # [r1, r2, r3, l]
MuonDetector_EndCap = [57.5, 78.2, 645, 563.8 - 417.9] # [r1, r2, r3, l]

####################################################################################################################
########################################## Computing the volume of every part ######################################
####################################################################################################################
vol_Nozzles1 = Vol("Nozzles", Nozzles1)
vol_Nozzles2 = Vol("Nozzles", Nozzles2)
vol_EcalBarrel = Vol("ECal", ECal_Barrel)
vol_EcalEndcap = Vol("ECal", ECal_Endcap)
vol_HcalBarrel = Vol("HCal", HCal_Barrel)
vol_HcalEndcap1 = Vol("HCal", HCal_Endcap1)
vol_HcalEndcap2 = Vol("HCal", HCal_Endcap2)
vol_SolenoidInner = Vol("Solenoid", Solenoid_Inner)
vol_SolenoidMiddle = Vol("Solenoid", Solenoid_Middle)
vol_SolenoidOuter =Vol("Solenoid", Solenoid_Outer)
vol_MuonDetectorBarrel = Vol("MuonDetector", MuonDetector_Barrel)
vol_MuonDetectorEndcap = Vol("MuonDetector", MuonDetector_EndCap)

det_vols = [vol_Nozzles1, vol_Nozzles2, vol_EcalBarrel, vol_EcalEndcap, vol_HcalBarrel, vol_HcalEndcap1, vol_HcalEndcap2,
            vol_SolenoidInner, vol_SolenoidMiddle, vol_SolenoidOuter, vol_MuonDetectorBarrel, vol_MuonDetectorEndcap]
####################################################################################################################
########################################## Printing the Tables with the names ######################################
####################################################################################################################
headers = ["Detector Parts", "Volume (m³)"]
det_names = ["Nozzles 1","Nozzles 2","ECal (Barrel)","ECal (Endcap)","HCal (Barrel)","HCal (Endcap 1)","Hcal (Endcap2)",
             "Solenoid (Inner)","Solenoid (Middle)","Soleonid (Outer)","Muon Detector (Barrel)","Muon Detector (Endcap)"]

data = []
for i in range(len(det_names)):
    data.append([])
    data[i] = [det_names[i],det_vols[i]]
print(tabulate(data, headers=headers, tablefmt="grid"))
```
