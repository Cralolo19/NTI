```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patheffects as pe

plt.style.use('sty.mplstyle')
```

# Altmannshofer Nucleon Form Factor

```python
mp = 0.938272088 # GeV
mn = 0.939565420 # GeV
mu_p = 2.793
mu_n = -1.913

def GpE_Alt(Q2):
    tau = Q2/(4*mp**2)
    num = 1 + (-0.19)*tau
    denom = 1 + (11.12)*tau + (15.16)*tau**2 + (21.25)*tau**3
    return num/denom

def GpM_Alt(Q2):
    tau = Q2/(4*mp**2)
    num = 1 + (1.09)*tau
    denom = 1 + (12.31)*tau + (25.57)*tau**2 + (30.61)*tau**3
    return (mu_p*num)/denom

def GnM_Alt(Q2):
    tau = Q2/(4*mn**2)
    num = 1 + (8.28)*tau
    denom = 1 + (21.3)*tau + (77)*tau**2 + (238)*tau**3
    return (mu_n*num)/denom

def GD_Alt(Q2):
    return 1/(1 + Q2/0.71)**2

def GnE_Alt(Q2):
    tau = Q2/(4*mn**2)
    num = (1.68)*tau*GD_Alt(Q2)
    denom = 1 + (3.63)*tau
    return num/denom
```

# Kelly Nucleon Form Factor

```python
def GpE_Kelly(Q2):
    tau = Q2/(4*mp**2)
    num = 1 + (-0.24)*tau
    denom = 1 + (10.98)*tau + (12.82)*tau**2 + (21.97)*tau**3
    return num/denom

def GpM_Kelly(Q2):
    tau = Q2/(4*mp**2)
    num = 1 + (0.12)*tau
    denom = 1 + (10.97)*tau + (18.86)*tau**2 + (6.55)*tau**3
    return (mu_p*num)/denom

def GnM_Kelly(Q2):
    tau = Q2/(4*mp**2)
    num = 1 + (2.33)*tau
    denom = 1 + (14.72)*tau + (24.20)*tau**2 + (84.1)*tau**3
    return (mu_n*num)/denom

def GD_Kelly(Q2):
    return 1/(1 + Q2/0.71)**2

def GnE_Kelly(Q2):
    tau = Q2/(4*mp**2)
    num = (1.70)*tau*GD_Kelly(Q2)
    denom = 1 + (3.30)*tau
    return num/denom
```

```python
Qmax = 3.0 # GeV
Qvals = np.linspace(0, Qmax, 500)
abs_values_GpE_Alt = np.abs( GpE_Alt(Qvals**2) )
abs_values_GpM_Alt = np.abs( GpM_Alt(Qvals**2) )
abs_values_GnM_Alt = np.abs( GnM_Alt(Qvals**2) )
abs_values_GnE_Alt = np.abs( GnE_Alt(Qvals**2) )
abs_values_GpE_Kelly = np.abs( GpE_Kelly(Qvals**2) )
abs_values_GpM_Kelly = np.abs( GpM_Kelly(Qvals**2) )
abs_values_GnM_Kelly = np.abs( GnM_Kelly(Qvals**2) )
abs_values_GnE_Kelly = np.abs( GnE_Kelly(Qvals**2) )
```

```python
fig = plt.figure()
ax = fig.add_subplot()

ax.plot(Qvals, abs_values_GpE_Alt, label=r'$G^p_E$ [Alt]')
ax.plot(Qvals, abs_values_GpM_Alt, label=r'$G^p_M$ [Alt]')
ax.plot(Qvals, abs_values_GpE_Kelly, label=r'$G^p_E$ [Kelly]')
ax.plot(Qvals, abs_values_GpM_Kelly, label=r'$G^p_M$ [Kelly]')

ax.set_yscale('log')
ax.set_xlim([0, Qmax])
ax.legend(fontsize=25, loc='upper right', frameon=True, framealpha=1, edgecolor='k', fancybox=True)
ax.tick_params(which='both', right=True, top=True)
ax.xaxis.grid(True, ls='-', which='major', color='gray', alpha=0.35)
ax.yaxis.grid(True, ls='-', which='both', color='gray', alpha=0.35)

ax.set_xlabel(r'Momentum Transfer $Q$ [GeV]')
ax.set_ylabel(r'$|G^p(Q^2)|$')
ax.set_title(r'Proton Form Factors')

plt.savefig('../Plots/Figures_PDF/proton_form_factor.pdf', transparent=False, bbox_inches='tight')
plt.savefig('../Plots/Figures_PNG/proton_form_factor.png', transparent=False, bbox_inches='tight')
```

```python
fig = plt.figure()
ax = fig.add_subplot()

ax.plot(Qvals, abs_values_GnE_Alt, label=r'$G^n_E$ [Alt]')
ax.plot(Qvals, abs_values_GnM_Alt, label=r'$G^n_M$ [Alt]')
ax.plot(Qvals, abs_values_GnE_Kelly, label=r'$G^n_E$ [Kelly]')
ax.plot(Qvals, abs_values_GnM_Kelly, label=r'$G^n_M$ [Kelly]')


ax.set_yscale('log')
ax.set_xlim([0, Qmax])
ax.legend(fontsize=25, loc='upper right', frameon=True, framealpha=1, edgecolor='k', fancybox=True)
ax.tick_params(which='both', right=True, top=True)
ax.xaxis.grid(True, ls='-', which='major', color='gray', alpha=0.35)
ax.yaxis.grid(True, ls='-', which='both', color='gray', alpha=0.35)

ax.set_xlabel(r'Momentum Transfer $Q$ [GeV]')
ax.set_ylabel(r'$|G^n(Q^2)|$')
ax.set_title(r'Neutron Form Factors')

plt.savefig('../Plots/Figures_PDF/neutron_form_factor.pdf', transparent=False, bbox_inches='tight')
plt.savefig('../Plots/Figures_PNG/neutron_form_factor.png', transparent=False, bbox_inches='tight')
```
