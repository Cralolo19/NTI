```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patheffects as pe
import matplotlib as mpl

plt.style.use('sty.mplstyle')
```

# MiniBooNE

```python
Form_Factors = ['3pF', 'Helm', 'KN', '(ad.) KN']
Number_FormFactors = len(Form_Factors)
Processes = [r'{\bf $\nu_\mu \to \nu_\mu \mu^{+} \mu_{-}$}', r'{\bf $\nu_\mu \to \nu_\mu e^{+} e_{-}$}', r'{\bf $\nu_\mu \to \nu_e e^{+} \mu_{-}$}']
Number_Processes = len(Processes)
fig = plt.figure(figsize=(12,14))
gs = fig.add_gridspec(Number_Processes, hspace=0)
ax1, ax2, ax3 = gs.subplots(sharex=True, sharey=True)
cols = ['rebeccapurple', 'teal', 'goldenrod']

Events_MiniBooNE_Process1 = np.array([0.01246, 0.01266, 0.01046, 0.01288], dtype=np.float64)
Events_MiniBooNE_Process2 = np.array([0.17742, 0.17828, 0.16700, 0.17865], dtype=np.float64)
Events_MiniBooNE_Process3 = np.array([0.27590, 0.27985, 0.25264, 0.28049], dtype=np.float64)

Uncertainty_Events_MiniBooNE_Process1 = np.array([0.0, 1.69, 16.02, 3.41], dtype=np.float64)
Uncertainty_Events_MiniBooNE_Process2 = np.array([0.0, 0.48, 4.19, 0.69], dtype=np.float64)
Uncertainty_Events_MiniBooNE_Process3 = np.array([0.0, 1.43, 8.43, 1.67], dtype=np.float64)

for ax, data, data_err, col in zip([ax1,ax2,ax3], [Events_MiniBooNE_Process1,Events_MiniBooNE_Process2,Events_MiniBooNE_Process3], [Uncertainty_Events_MiniBooNE_Process1, Uncertainty_Events_MiniBooNE_Process2, Uncertainty_Events_MiniBooNE_Process3], cols):
    ax.barh(np.flipud(Form_Factors), np.flipud(data), left=0, align='center', height=1, edgecolor='k', facecolor='w', lw=1.5, zorder=-10, alpha=1)
    ax.barh(np.flipud(Form_Factors), np.flipud(data), left=0, xerr=0, align='center', height=0.98, facecolor=col, lw=0, zorder=-10, alpha=0.5)
    ax.patch.set_facecolor(col)
    ax.patch.set_alpha(0.1)

# Adding text labels
t_shift = [0.01, 0.01, 0.01, 0.0, 0.0]
for i in range(Number_FormFactors):
    ax1.text(0.08 + Events_MiniBooNE_Process1[i], 3-i-t_shift[i], r'{\bf %.3f [%.2f $\%%$]}' % (Events_MiniBooNE_Process1[i], Uncertainty_Events_MiniBooNE_Process1[i]), va='center', ha='right', fontsize=20)
    ax2.text(0.95*Events_MiniBooNE_Process2[i], 3-i-t_shift[i], r'{\bf %.3f [%.2f $\%%$]}' % (Events_MiniBooNE_Process2[i], Uncertainty_Events_MiniBooNE_Process2[i]), va='center', ha='right', fontsize=20)
    ax3.text(0.97*Events_MiniBooNE_Process3[i], 3-i-t_shift[i], r'{\bf %.3f [%.2f $\%%$]}' % (Events_MiniBooNE_Process3[i], Uncertainty_Events_MiniBooNE_Process3[i]), va='center', ha='right', fontsize=20)

# Grid lines
ax1.xaxis.grid(True, linestyle='--', which='major', color='grey', alpha=0.55)
ax2.xaxis.grid(True, linestyle='--', which='major', color='grey', alpha=0.55)
ax3.xaxis.grid(True, linestyle='--', which='major', color='grey', alpha=0.55)


# Tick parameters
ax1.tick_params(which='both', right=False, tickdir='out', top=False, bottom=False)
ax2.tick_params(which='both', right=False, tickdir='out', top=False, bottom=False)
ax3.tick_params(which='both', right=False, tickdir='out', top=False)

# Labels
ax1.text(0.39, 0.0, Processes[0], fontsize=30, alpha=0.7, color=cols[0], ha='right', rotation=0, path_effects=[pe.Stroke(linewidth=1.5, foreground='k', alpha=0.5), pe.Normal()])
ax2.text(0.39, 0.0, Processes[1], fontsize=30, alpha=0.7, color=cols[1], ha='right', rotation=0, path_effects=[pe.Stroke(linewidth=1.5, foreground='k', alpha=0.5), pe.Normal()])
ax3.text(0.39, 0.0, Processes[2], fontsize=30, alpha=0.7, color=cols[2], ha='right', rotation=0, path_effects=[pe.Stroke(linewidth=1.5, foreground='k', alpha=0.5), pe.Normal()])

ax1.set_title(r'Number of Events at MiniBooNE')
ax3.set_xlabel(r'Number of Trident Events')
ax3.set_xlim(right=0.4)
plt.savefig('../Plots/Figures_PDF/EventsatMiniBooNE.pdf', transparent=False, bbox_inches='tight')
plt.savefig('../Plots/Figures_PNG/EventsatMiniBooNE.png', transparent=False, bbox_inches='tight')
```
