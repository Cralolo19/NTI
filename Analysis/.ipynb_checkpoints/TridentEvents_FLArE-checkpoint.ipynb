```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import simpson

plt.style.use('sty.mplstyle')

def ComputeIntegral(x1, x2, x, y):
    # x1/x2 is the starting/ending point of the integral
    # x/y represents our data (points) in x/y axis
    cond = (x >= x1) & (x <= x2)
    y1 = np.interp(x1, x, y)
    y2 = np.interp(x2, x, y)
    xval = np.concatenate(([x1], x[cond], [x2]))
    yval = np.concatenate(([y1], y[cond], [y2]))
    integral = simpson(yval, x=xval)
    return integral

vmu_N = np.array([4121433365878.5957, 5018202663720.177, 5817899388961.731, 7083465994613.952, 8211769435882.73,
                  9519796850774.973, 10253098844537.88, 11042886477909.086, 11324561586429.418, 11331094275479.219,
                  11619758585756.848, 11345055065027.967, 10546305846177.21, 10047595523814.543, 9113840155727.848,
                  8267248236162.711, 7317556069183.182, 6319796750300.388, 5196823325644.494, 3873896954673.7803,
                  2817714046574.8394, 1927612340370.7134, 1110591330873.485, 545554395361.5635, 248980387104.4981,
                  103008782978.99939, 18058273266.98156, 136992465.719387]) # nu's
vmu_E = np.logspace(1,4,31)[0:len(vmu_N)+1]
vmu_Emin = vmu_E[0:-1]
vmu_Emax = vmu_E[1:]

#vmu_dN_dE = vmu_N/ (vmu_Emax - vmu_Emin)
#vmu_dN_dE *= (vmu_Emax - vmu_Emin) / sum(vmu_N)
Lum = 3000
vmu_N = vmu_N/sum(vmu_N)

data_vmu_to_vmu_mu_mu = np.loadtxt('../csv/CrossSections/vmu_to_vmu_mu+_mu-_xsec/Coherent/Argon/3pF/vmu_to_vmu_mu+_mu-_coh_Ar_[3pF]_xsec', delimiter=',')
vmu_to_vmu_mu_mu_Enu = data_vmu_to_vmu_mu_mu[:,0]
vmu_to_vmu_mu_mu_Xsec = data_vmu_to_vmu_mu_mu[:,1]
vmu_to_vmu_mu_mu_errXsec = data_vmu_to_vmu_mu_mu[:,2]

NumOfEvents = 0
for i in range(len(vmu_N)):
    Xsec_int = ComputeIntegral(x1=vmu_Emin[i], x2=vmu_Emax[i], x=vmu_to_vmu_mu_mu_Enu, y=vmu_to_vmu_mu_mu_Xsec)
    mean_Xsec = Xsec_int/(vmu_Emax[i] - vmu_Emin[i])
    NumOfEvents += vmu_N[i] * mean_Xsec
NumOfEvents *= Lum
NumOfEvents
```
