```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patheffects as pe
import matplotlib

plt.style.use('sty.mplstyle')
```

```python
Energy_NuMI = np.loadtxt('../arXiv-2504.08950v1/Digitized_Dataset_NuMI_offaxis_numu_nuMode.txt', delimiter=',', usecols=0)
numu_Flux_off_axis_NuMI = np.loadtxt('../arXiv-2504.08950v1/Digitized_Dataset_NuMI_offaxis_numu_nuMode.txt', delimiter=',', usecols=1)

```

```python
fig = plt.figure()
ax = fig.add_subplot()

ax.plot(Energy_NuMI, numu_Flux_off_axis_NuMI)

ax.set_yscale('log')
ax.set_ylim([2e-12, 3e-5])
ax.set_xlim([0,6])

ax.set_title(r'{\bf NuMI} Simulation (Off-axis) [Digitized Data]')
ax.set_ylabel(r'$\Phi_\nu$/GeV/POT')
ax.set_xlabel(r'Neutrino Energy $E_\nu$ [GeV]')
```
