```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patheffects as pe
from scipy.integrate import simpson
import csv

plt.style.use('sty.mplstyle')
```

# BNB flux at MiniBooNE ($\nu$ Mode)

```python
# Upload the Data (Fluxes Should be multiplied by 2x10^5)
Elow_MiniBooNE_nuMode = np.loadtxt("../csv/Fluxes/MiniBooNE_Fluxes_NuMode.dat.txt", usecols=0, comments="#") # Lower bin edge in GeV
Ehigh_MiniBooNE_nuMode = np.loadtxt("../csv/Fluxes/MiniBooNE_Fluxes_NuMode.dat.txt", usecols=1, comments="#") # Upper bin edge in GeV
numu_flux_MiniBooNE_nuMode = 2e5*np.loadtxt("../csv/Fluxes/MiniBooNE_Fluxes_NuMode.dat.txt", usecols=2, comments="#") # numu/m^2/POT/GeV
numubar_flux_MiniBooNE_nuMode = 2e5*np.loadtxt("../csv/Fluxes/MiniBooNE_Fluxes_NuMode.dat.txt", usecols=3, comments="#") # numubar/m^2/POT/GeV

Energy_MiniBooNE_nuMode = (Elow_MiniBooNE_nuMode + Ehigh_MiniBooNE_nuMode)/2

# Computing the Normalized Probabilities
nu_total_flux_MiniBooNE_nuMode = simpson(numu_flux_MiniBooNE_nuMode, Elow_MiniBooNE_nuMode)
antinu_total_flux_MiniBooNE_nuMode = simpson(numubar_flux_MiniBooNE_nuMode, Elow_MiniBooNE_nuMode)
nu_normProb_MiniBooNE_nuMode = np.zeros_like(numu_flux_MiniBooNE_nuMode)
antinu_normProb_MiniBooNE_nuMode = np.zeros_like(numubar_flux_MiniBooNE_nuMode)

for i in range(len(numu_flux_MiniBooNE_nuMode)):
    bin_width = Ehigh_MiniBooNE_nuMode[i] - Elow_MiniBooNE_nuMode[i]
    nu_normProb_MiniBooNE_nuMode[i] = numu_flux_MiniBooNE_nuMode[i]*bin_width/nu_total_flux_MiniBooNE_nuMode

for i in range(len(numubar_flux_MiniBooNE_nuMode)):
    bin_width = Ehigh_MiniBooNE_nuMode[i] - Elow_MiniBooNE_nuMode[i]
    antinu_normProb_MiniBooNE_nuMode[i] = numubar_flux_MiniBooNE_nuMode[i]*bin_width/antinu_total_flux_MiniBooNE_nuMode


# Writing to CSV
with open("../csv/Fluxes/NormProb/nu_NormProb_BNB_MiniBooNE_nuMode.txt", mode='w', newline="") as file:
    file.write("########################################################################################## \n")
    file.write(f"#####\t Integrated Neutrino Flux (Neutrino Mode): {nu_total_flux_MiniBooNE_nuMode} m^-2 POT^-1 \n")
    file.write("########################################################################################## \n")
    for e_low, e_high, p_norm in (zip(Elow_MiniBooNE_nuMode, Ehigh_MiniBooNE_nuMode, nu_normProb_MiniBooNE_nuMode)):
        file.write(f"\t {e_low:.4f}, \t {e_high:.4f}, \t {p_norm:.22f}, \n")

with open("../csv/Fluxes/NormProb/antinu_NormProb_BNB_MiniBooNE_nuMode.txt", mode='w', newline="") as file:
    file.write("########################################################################################## \n")
    file.write(f"#####\t Integrated Antineutrino Flux (Neutrino Mode): {antinu_total_flux_MiniBooNE_nuMode} m^-2 POT^-1 \n")
    file.write("########################################################################################## \n")
    for e_low, e_high, p_norm in (zip(Elow_MiniBooNE_nuMode, Ehigh_MiniBooNE_nuMode, antinu_normProb_MiniBooNE_nuMode)):
        file.write(f"\t {e_low:.4f}, \t {e_high:.4f}, \t {p_norm:.22f}, \n")
```

# Plotting the Fluxes

```python
fig = plt.figure()
ax = fig.add_subplot()

ax.plot(Energy_MiniBooNE_nuMode, numu_flux_MiniBooNE_nuMode, ls='-', color='purple', label=r'$\nu_\mu$ [$\nu$-Mode]', path_effects=[pe.Stroke(linewidth=4, foreground='k'), pe.Normal()])
ax.plot(Energy_MiniBooNE_nuMode, numubar_flux_MiniBooNE_nuMode, ls='--', color='purple', label=r'$\overline{\nu}_\mu$ [$\nu$-Mode]', path_effects=[pe.Stroke(linewidth=4, foreground='k'), pe.Normal()])
ax.plot(Energy_MiniBooNE_nuMode, numu_flux_MiniBooNE_antinuMode, ls='-', color='orange', label=r'$\nu_\mu$ [$\overline{\nu}$-Mode]', path_effects=[pe.Stroke(linewidth=4, foreground='k'), pe.Normal()])
ax.plot(Energy_MiniBooNE_nuMode, numubar_flux_MiniBooNE_antinuMode, ls='--', color='orange', label=r'$\overline{\nu}_\mu$ [$\overline{\nu}$-Mode]', path_effects=[pe.Stroke(linewidth=4, foreground='k'), pe.Normal()])

ax.set_xlim([0,7.5])
ax.set_yscale('log')
ax.set_ylim(top=1e-5)
ax.xaxis.grid(True, ls='-', which='major', color='gray', alpha=0.35)
ax.yaxis.grid(True, ls='-', which='major', color='gray', alpha=0.35)
ax.tick_params(which='both', right=False, top=False)
ax.set_title(r'MiniBooNE Flux - $\nu_\mu$ and $\overline{\nu}_\mu$', pad=16)
ax.set_xlabel(r'Neutrino Energy $E_\nu$ [GeV]')
ax.set_ylabel(r'$\frac{\mathrm{d} \Phi}{\mathrm{d} E_\nu}$  [$\nu$/m$^2$/POT/GeV]')

ax.legend(fontsize=25, loc='upper right', frameon=False, framealpha=1, edgecolor='k', fancybox=True)

plt.savefig('../Plots/Figures_PDF/MiniBooNE_Flux.pdf', transparent=False, bbox_inches='tight')
plt.savefig('../Plots/Figures_PNG/MiniBooNE_Flux.png', transparent=False, bbox_inches='tight')
```

```python
Energy_DUNEStd_nuMode = np.loadtxt("../csv/Fluxes/histos_g4lbne_v3r5p4_QGSP_BERT_OptimizedEngineeredNov2017_neutrino_LBNEND_globes_flux.txt", usecols=0, comments="#") # Middle bin edge in GeV
numu_flux_DUNEStd_nuMode = np.loadtxt("../csv/Fluxes/histos_g4lbne_v3r5p4_QGSP_BERT_OptimizedEngineeredNov2017_neutrino_LBNEND_globes_flux.txt", usecols=2, comments="#") # numu/m^2/POT/GeV

Elow_DUNEStd_nuMode = Energy_DUNEStd_nuMode - 0.25/2
Ehigh_DUNEStd_nuMode = Energy_DUNEStd_nuMode + 0.25/2


# Computing the Normalized Probabilities
nu_total_flux_DUNEStd_nuMode = simpson(numu_flux_DUNEStd_nuMode, Elow_DUNEStd_nuMode)
nu_normProb_DUNEStd_nuMode = np.zeros_like(numu_flux_DUNEStd_nuMode)

for i in range(len(numu_flux_DUNEStd_nuMode)):
    bin_width = Ehigh_DUNEStd_nuMode[i] - Elow_DUNEStd_nuMode[i]
    nu_normProb_DUNEStd_nuMode[i] = numu_flux_DUNEStd_nuMode[i]*bin_width/nu_total_flux_DUNEStd_nuMode

# Writing to CSV
with open("../csv/Fluxes/NormProb/nu_NormProb_DUNEStd_nuMode.txt", mode='w', newline="") as file:
    file.write("########################################################################################## \n")
    file.write(f"#####\t Integrated Neutrino Flux (Neutrino Mode): {nu_total_flux_DUNEStd_nuMode} m^-2 POT^-1 \n")
    file.write("########################################################################################## \n")
    for e_low, e_high, p_norm in (zip(Elow_DUNEStd_nuMode, Ehigh_DUNEStd_nuMode, nu_normProb_DUNEStd_nuMode)):
        file.write(f"\t {e_low:.4f}, \t {e_high:.4f}, \t {p_norm:.22f}, \n")
```
