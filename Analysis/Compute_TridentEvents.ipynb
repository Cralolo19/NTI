```python
import numpy as np

# Define the constants
N_POT = 3.3e21 # Protons On Target
F_mu = 1.148e-3 # m^{-2} POT^{-1}
M_D = 67*(1000) # Detector mass Kg
M_Ar = 39.948*(1.66053906660e-27) # Kg
Conversion = 1e-43 # from fb to  m^{2}

def Trident_Events(X_section, X_section_ref):
    aux = Conversion*X_section*N_POT*M_D/M_Ar
    N_trident = F_mu*aux
    error = 1 - X_section/X_section_ref
    error = 100*error
    error = np.abs(error)
    print("Number of Trident Events \t Uncertainty")
    for i in range(len(X_section)):
        print(N_trident[i], "\t\t", f"[{error[i]:.2f}]")
```
