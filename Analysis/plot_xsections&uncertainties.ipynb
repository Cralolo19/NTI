# Uncertainties on the $\nu_\mu \to \nu_e e^{+} \mu^{-}$ process

```python
def NoNull_Xsec(Xsec, Energy):
    new_Xsec = Xsec[Xsec!=0]
    new_Energy = Energy[Xsec!=0]
    return new_Energy, new_Xsec

New_Energy_ve_e_mu_coh_Ar_3pF, New_Xsec_ve_e_mu_coh_Ar_3pF = NoNull_Xsec(Xsec_ve_e_mu_coh_Ar_3pF, Energy_ve_e_mu_coh_Ar_3pF)
New_Energy_ve_e_mu_coh_Ar_Helm, New_Xsec_ve_e_mu_coh_Ar_Helm = NoNull_Xsec(Xsec_ve_e_mu_coh_Ar_Helm, Energy_ve_e_mu_coh_Ar_Helm)
New_Energy_ve_e_mu_coh_Ar_KN, New_Xsec_ve_e_mu_coh_Ar_KN = NoNull_Xsec(Xsec_ve_e_mu_coh_Ar_KN, Energy_ve_e_mu_coh_Ar_KN)
New_Energy_ve_e_mu_coh_Ar_adKN, New_Xsec_ve_e_mu_coh_Ar_adKN = NoNull_Xsec(Xsec_ve_e_mu_coh_Ar_adKN, Energy_ve_e_mu_coh_Ar_adKN)

Uncertainty_ve_e_mu_coh_Ar_Helm = 100*np.abs(1 - New_Xsec_ve_e_mu_coh_Ar_Helm/New_Xsec_ve_e_mu_coh_Ar_3pF)
Uncertainty_ve_e_mu_coh_Ar_KN = 100*np.abs(1 - New_Xsec_ve_e_mu_coh_Ar_KN/New_Xsec_ve_e_mu_coh_Ar_3pF)
Uncertainty_ve_e_mu_coh_Ar_adKN = 100*np.abs(1 - New_Xsec_ve_e_mu_coh_Ar_adKN/New_Xsec_ve_e_mu_coh_Ar_3pF)

fig = plt.figure()
ax = fig.add_subplot()

ax.plot(New_Energy_ve_e_mu_coh_Ar_3pF, Uncertainty_ve_e_mu_coh_Ar_Helm, label=r'Helm', lw=3, path_effects=[pe.Stroke(linewidth=4, foreground='k'), pe.Normal()])
ax.plot(New_Energy_ve_e_mu_coh_Ar_3pF, Uncertainty_ve_e_mu_coh_Ar_KN, label=r'KN', lw=3, path_effects=[pe.Stroke(linewidth=4, foreground='k'), pe.Normal()])
ax.plot(New_Energy_ve_e_mu_coh_Ar_3pF, Uncertainty_ve_e_mu_coh_Ar_adKN, label=r'(ad.) KN', lw=3, path_effects=[pe.Stroke(linewidth=4, foreground='k'), pe.Normal()])

ax.set_xscale('log')
ax.set_ylim([0, 18])
ax.set_xlim([1e-1,120])

ax.xaxis.grid(True, which='major', ls=ls[0], alpha=0.35)
ax.yaxis.grid(True, which='major', ls=ls[0], alpha=0.35)

ax.set_title(r'$\nu_\mu \to \nu_e e^{+} \mu^{-}$ [Argon $^{40}$Ar]')
ax.set_ylabel(r'$\Delta \sigma (E_\nu) [\%]$')
ax.set_xlabel(r'Neutrino Energy $E_\nu$ [GeV]')
ax.legend(fontsize=25, frameon=True, framealpha=1, edgecolor='k', fancybox=True)

plt.savefig('../Plots/Figures_PDF/40Argon_Uncertainty_ve_e_mu_coh.pdf', transparent=False, bbox_inches='tight')
plt.savefig('../Plots/Figures_PNG/40Argon_Uncertainty_ve_e_mu_coh.png', transparent=False, bbox_inches='tight')
```
