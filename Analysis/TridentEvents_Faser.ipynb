```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patheffects as pe
import matplotlib

plt.style.use('sty.mplstyle')
```

```python
Enu_FASER_WS = np.loadtxt('../csv/CrossSections/vmu_to_vmu_mu+_mu-_xsec/Coherent/Tungsten/WoodsSaxon/vmu_to_vmu_mu+_mu-_coh_W_[WS]_xsec', usecols=0, delimiter=',') #GeV
Xsec_FASER_WS = np.loadtxt('../csv/CrossSections/vmu_to_vmu_mu+_mu-_xsec/Coherent/Tungsten/WoodsSaxon/vmu_to_vmu_mu+_mu-_coh_W_[WS]_xsec', usecols=1, delimiter=',') #fb
Xsec_FASER_WS *= 1e-39 # cm^2
```

```python
Enu_FASER_Helmv1 = np.loadtxt('../csv/CrossSections/vmu_to_vmu_mu+_mu-_xsec/Coherent/Tungsten/Helmv1/vmu_to_vmu_mu+_mu-_coh_W_[H1]_xsec', usecols=0, delimiter=',') #GeV
Xsec_FASER_Helmv1 = np.loadtxt('../csv/CrossSections/vmu_to_vmu_mu+_mu-_xsec/Coherent/Tungsten/Helmv1/vmu_to_vmu_mu+_mu-_coh_W_[H1]_xsec', usecols=1, delimiter=',') #fb
Xsec_FASER_Helmv1 *= 1e-39 # cm^2
```

```python
fig = plt.figure()
ax = fig.add_subplot()

ax.plot(Enu_FASER_WS, Xsec_FASER_WS/Enu_FASER_WS, label='WS FF', path_effects=[pe.Stroke(linewidth=5, foreground='k'), pe.Normal()])
ax.plot(Enu_FASER_Helmv1, Xsec_FASER_Helmv1/Enu_FASER_Helmv1, label='Helmv1 FF', path_effects=[pe.Stroke(linewidth=5, foreground='k'), pe.Normal()])

ax.set_yscale('log')
ax.set_xscale('log')
ax.set_xlim([1e-1,1e4])
ax.set_ylim([1e-56, 3e-36])

ax.legend(fontsize=20, loc='upper right', frameon=True, framealpha=1, edgecolor='k', fancybox=True)
ax.set_xlabel(r'Neutrino Energy $E_\nu$ [GeV]')
ax.set_ylabel(r'Cross Section/$E_\nu$ [cm$^2$ GeV$^{-1}$]')
ax.set_title(r'Tungsten $^{184}$W')
ax.xaxis.grid(True, which='major', ls='-', alpha=0.35)
ax.yaxis.grid(True, which='major', ls='-', alpha=0.35)

plt.savefig('../Plots/Figures_PDF/184Tungsten_Xsection.pdf', transparent=False, bbox_inches='tight')
plt.savefig('../Plots/Figures_PNG/184Tungsten_Xsection.png', transparent=False, bbox_inches='tight')
```
